```python
my_List = [1,4,2,6]
arr = np.array(my_List)
print(arr)
# this is the way to create array in Python by making use of list data structure
```

```python
# this is the way to create 2D array
my_lst = [2,5,3,1]
my_lst2 = [5,0,1,3]
my_lst3 = [3,6,9,1]

arr2 = np.array([my_lst,my_lst2,my_lst3])
print(arr2)
```

```python
# We can also chack the size or shape of the array

arr2.shape
# this shows whats the size of the array ie. it has 3 rows and 4 cols
```

```python
# we can also reshape it by using reshape inbuilt function

arr2.reshape(6,2)

# total number of elements should not exceed 12 because 3x4 this will give you total number of elements
```

```python
# arange function inbuilt function
arr = np.arange(0,21)
arr
# using this we are able to get all the numbers between 0 to 20.
```

```python
# arange function has 3 parameters 1) start 2) stop 3) step
# step - tells us how many elements to skip till the end
arr = np.arange(0,31,2)
arr
```

```python
# Almost same as arange only diff is it has 3rd paramter where it define how many division you want.

np.linspace(1,5, 20)
```
