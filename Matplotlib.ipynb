```python
import numpy as np
import pandas as pd
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
# using this line there is no need to write plt.show() to show graph


# this is a visualization library used to plot graphs or represent 2D/3D diagram
# Its base library is Numpy (Numerical Extension of Numpy)
```

```python
x = np.arange(0,10)
y = np.arange(11,21)


# to show on graph we need atleast 2 variables
```

```python
# plotting using matplotlib

# scatter plot - used to scatter the values of x and y variables

plt.scatter(x,y, c='r')
```

```python
# We can also give name to our x axis and y axis

plt.scatter(x,y, c='r')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
```

```python
# Also we can provide Title to our Grapgh

plt.scatter(x,y, c='r')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.title('Graph in 2D')
```

```python
# This is a plot of the x and y values this is different from scatter plot


plt.plot(x,y,'g--')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
```

```python
# We can also subplot i.e we can create small small graphs

# subplot takes 3 parameters 1) no. of rows 2) no. of cols 3) Position of the graph

# plt.subplot(2,2,1) in this there are 2 rows and 2 cols and 1 indicates the position of the diagram

plt.subplot(2,2,1)
plt.plot(x,y,'g*')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')

plt.subplot(2,2,2)
plt.plot(x,y,'r*-')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')

plt.subplot(2,2,3)
plt.plot(x,y,'b--')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')

plt.subplot(2,2,4)
plt.plot(x,y,'c--')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
```

```python
# you can aslo plot sine curves, cosine curves by using their respective functions

x = np.arange(0, 4*np.pi, 0.1)
y = np.sin(x) # these are the sine and cosine functions and we are passing x value in it.
z = np.cos(x)

plt.plot(x,y)
plt.plot(x,z)
```

```python
x = np.arange(0, 6 * np.pi, 0.1)

y = np.cos(x)

plt.plot(x,y, 'r--')
```
