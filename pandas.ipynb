```python
import numpy as np
import pandas as pd
```

```python
# pd.DataFrame is a function that takes 1) Data 2) Index for rows 3) for columns

df = pd.DataFrame(np.arange(0,30).reshape(5,6), index=['Row1','Row2','Row3','Row4','Row5'], columns=['Column1','Column2','Column3','Column4','Column5','Column6'])
```

```python
df.head()
```

```python
df.loc['Row1']

# this function focuses on rows and shows all the data related to the given row
```

```python
# iloc[ ] function - In this as we used to give rows and column for accessing the 2D array elements in similar way we have to give here.

df.iloc[:,:]

# this will show all the rows and columns and if we want particular part form this we can aslo do that by mentions till what row or col we want.
```

```python
# We can also convert our DataFrame into arrays

df.iloc[:,:].values

# By using iloc function we can convert it into Arrays.
```

```python
# To check our DataFrame contains Null values We can make use of the isNull funvtion and Sum function with our DATAFRAME

df.isnull().sum()

# now out Data does not contain any Null values so it should show 0
```

```python
# String IO function

from io import StringIO, BytesIO
```

```python
data = ('col1,col2,col3\n'
        'x, y, 1\n'
        'a, b, 2\n'
        'c, d, 3')
```

```python

# Using StringIO function you are actually reading raw/hard coded data and creating it into DataFrame.

pd.read_csv(StringIO(data))
```

```python
# Also if we want particular columns then we can do that as well

df = pd.read_csv(StringIO(data), usecols=['col1','col3'], skipinitialspace=True)
```

```python
df.head()

```

```python
# To access the particular element from the data


df['col1']
```

```python
df['col1'][2]
```
